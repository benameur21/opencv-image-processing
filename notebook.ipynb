Add Salt-and-Pepper Noise

```python
# Add salt-and-pepper noise

noise_1 = random_noise(gray, mode="s&p", amount=0.01)
noise_5 = random_noise(gray, mode="s&p", amount=0.05)
noise_10 = random_noise(gray, mode="s&p", amount=0.10)

# Convert values from [0, 1] to [0, 255]
noise_1 = (noise_1 * 255).astype(np.uint8)
noise_5 = (noise_5 * 255).astype(np.uint8)
noise_10 = (noise_10 * 255).astype(np.uint8)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(noise_1, cmap="gray")
plt.title("Salt & Pepper Noise - 1%")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(noise_5, cmap="gray")
plt.title("Salt & Pepper Noise - 5%")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(noise_10, cmap="gray")
plt.title("Salt & Pepper Noise - 10%")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 4, 1)
plt.imshow(gray, cmap="gray")
plt.title("Original Grayscale")
plt.axis("off")

plt.subplot(1, 4, 2)
plt.imshow(noise_1, cmap="gray")
plt.title("1% Noise")
plt.axis("off")

plt.subplot(1, 4, 3)
plt.imshow(noise_5, cmap="gray")
plt.title("5% Noise")
plt.axis("off")

plt.subplot(1, 4, 4)
plt.imshow(noise_10, cmap="gray")
plt.title("10% Noise")
plt.axis("off")

plt.tight_layout()
plt.show()
```

Create a Structuring Element

```python
# Rectangular structuring elements

kernel_rect_3 = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (3, 3)
)

kernel_rect_5 = cv2.getStructuringElement(
    cv2.MORPH_RECT,
    (5, 5)
)
```

```python
# Elliptical structuring elements

kernel_ellipse_3 = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (3, 3)
)

kernel_ellipse_5 = cv2.getStructuringElement(
    cv2.MORPH_ELLIPSE,
    (5, 5)
)
```

```python
plt.figure(figsize=(12, 6))

plt.subplot(2, 2, 1)
plt.imshow(kernel_rect_3, cmap="gray")
plt.title("Rectangle 3x3")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(kernel_rect_5, cmap="gray")
plt.title("Rectangle 5x5")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(kernel_ellipse_3, cmap="gray")
plt.title("Ellipse 3x3")
plt.axis("off")

plt.subplot(2, 2, 4)
plt.imshow(kernel_ellipse_5, cmap="gray")
plt.title("Ellipse 5x5")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python
print("Rectangle 3x3:")
print(kernel_rect_3)

print("\nRectangle 5x5:")
print(kernel_rect_5)

print("\nEllipse 3x3:")
print(kernel_ellipse_3)

print("\nEllipse 5x5:")
print(kernel_ellipse_5)
```

Apply Erosion

```python
# Erosion with different kernel sizes

erosion_3 = cv2.erode(
    noise_5,
    kernel_rect_3,
    iterations=1
)

erosion_5 = cv2.erode(
    noise_5,
    kernel_rect_5,
    iterations=1
)
```

```python
plt.figure(figsize=(12, 8))

plt.subplot(2, 2, 1)
plt.imshow(noise_5, cmap="gray")
plt.title("Noisy Image - 5%")
plt.axis("off")

plt.subplot(2, 2, 2)
plt.imshow(erosion_3, cmap="gray")
plt.title("Erosion - 3x3 - 1 iteration")
plt.axis("off")

plt.subplot(2, 2, 3)
plt.imshow(erosion_5, cmap="gray")
plt.title("Erosion - 5x5 - 1 iteration")
plt.axis("off")

plt.tight_layout()
plt.show()
```

```python

erosion_iter_1 = cv2.erode(
    noise_5,
    kernel_rect_3,
    iterations=1
)

erosion_iter_2 = cv2.erode(
    noise_5,
    kernel_rect_3,
    iterations=2
)

erosion_iter_3 = cv2.erode(
    noise_5,
    kernel_rect_3,
    iterations=3
)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(erosion_iter_1, cmap="gray")
plt.title("Erosion - 1 iteration")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(erosion_iter_2, cmap="gray")
plt.title("Erosion - 2 iterations")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(erosion_iter_3, cmap="gray")
plt.title("Erosion - 3 iterations")
plt.axis("off")

plt.tight_layout()
plt.show()
```
