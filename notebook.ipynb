Extract a channel directly

```python
B = image[:, :, 0]
G = image[:, :, 1]
R = image[:, :, 2]
```

Reconstruct the original image

```python
image_reconstructed = cv2.merge([B, G, R])
```

```python
plt.imshow(cv2.cvtColor(image_reconstructed, cv2.COLOR_BGR2RGB))
plt.axis("off")
plt.show()
```

Convert the Image to Grayscale

```python
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
```

```python
plt.imshow(gray, cmap="gray")
plt.title("Lena - Grayscale")
plt.axis("off")
plt.show()
```

```python
print("Grayscale dimensions:", gray.shape)
```

```python
print("Pixel at (100, 100):", gray[100, 100])
print("Pixel at (250, 250):", gray[250, 250])
print("Pixel at (400, 400):", gray[400, 400])
```

Transform the Image

```python
cv2.resize(image, (width, height))
```

```python
lena_256 = cv2.resize(image, (256, 256))
lena_128 = cv2.resize(image, (128, 128))
lena_800 = cv2.resize(image, (800, 800))
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(lena_256, cv2.COLOR_BGR2RGB))
plt.title("256 x 256")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(lena_128, cv2.COLOR_BGR2RGB))
plt.title("128 x 128")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(lena_800, cv2.COLOR_BGR2RGB))
plt.title("800 x 800")
plt.axis("off")

plt.show()
```

```python
# Vertical flip
flip_vertical = cv2.flip(image, 0)

# Horizontal flip
flip_horizontal = cv2.flip(image, 1)

# Horizontal + vertical flip
flip_both = cv2.flip(image, -1)
```

```python
plt.figure(figsize=(12, 4))

plt.subplot(1, 3, 1)
plt.imshow(cv2.cvtColor(flip_vertical, cv2.COLOR_BGR2RGB))
plt.title("Vertical Flip")
plt.axis("off")

plt.subplot(1, 3, 2)
plt.imshow(cv2.cvtColor(flip_horizontal, cv2.COLOR_BGR2RGB))
plt.title("Horizontal Flip")
plt.axis("off")

plt.subplot(1, 3, 3)
plt.imshow(cv2.cvtColor(flip_both, cv2.COLOR_BGR2RGB))
plt.title("Both Flips")
plt.axis("off")

plt.show()
```
